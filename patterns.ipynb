# Pattern Square

```python
n=5
for i in range(n):
    print([i for i in range(n)])
```

```python
for i in range(n):
    print("*"*n)
```

```python
# ABCD
# ABCD
for i in range(n):
    print([chr(i) for i in range(65,65+n)])
```

```python
# "123"
# "456"
# "789"
num=0
for i in range(n):
    for j in range(1,n+1):
        num+=1
        print(num,end=" ")
    print(end="\n")
          
```

```python
# "ABC"
# "DEF"
# "GHI"
num=0
for i in range(n):
    for j in range(1,n+1):
        num+=1
        print(chr(num+64),end=" ")
    print(end="\n")    
```

```python
# "*"
# "**"
# "***"
for i in range(1,n+1):
    print("*"*i)
```

```python
# "1"
# "22"
# "333"
for i in range(1,n+1):
    print(str(i)*i)
```

```python
# "A"
# "BB"
# "CCC"
for i in range(1,n+1):
    print(chr(i+64)*i)
```

```python
# "1"
# "12"
# "123"
arr=[i for i in range(1,n+1)]
# print(arr)
for i in range(1,n+1):
    print(arr[0:i])
```

```python
# "1"
# "21"
# "321"
arr=[i for i in range(1,n+1)]
# print(arr)
for i in range(1,n+1):
    print(arr[0:i][::-1])
```

# Floyds Triangle

```python
# Floyds Triangle
# "1"
# "23"
# "456"
num=1
for i in range(1,n+1):
    for j in range(1,i+1):
        print(num,end=" ")
        num+=1
    print(end="\n")
```
